### - Análise bivariada focada no alvo 


#### Este é o coração da EDA. 

- Comparamos cada variável explicativa com acidentes_fatais, sempre relatando volume e proporção - nunca só uma das duas


```python
#### Criação de uma variável-alvo binária: acidente com vítimas fatais (1) ou não (0) 
prf_acidentes_2025["acidente_fatal"] = prf_acidentes_2025["classificacao_acidente"].fillna("").apply(
    lambda x: 1 if any(p in x.lower() for p in ["fatais", "fatal", "óbito", "morte"]) else 0
)
```

- Definição da função de análise bivariada adaptada para a base da PRF
*** var: variável explicativa (ex: "uf", "causa_acidente" etc.) 
*** min_acidentes: mínimo de acidentes para inclusão na análise (evita categorias irrelevantes) 

```python
def analise_bivariada_prf(prf_acidentes_2025, var, min_acidentes=1): 
    
    tab = prf_acidentes_2025.groupby(var).agg(
        
        # total de acidentes na categoria
        acidentes_total=("id", "count"), 
        
        # total de acidentes com vítimas fatais na categoria
        acidentes_fatais=("acidente_fatal", "sum"), 
        
        # total de mortes na categoria 
        mortos=("mortos", "sum"), 
        
        # total de feridos na categoria
        feridos=("feridos", "sum")
    )
    
    ### proporção condicional de acidente fatal
    tab["pct_acidente_fatal"] = tab["acidentes_fatais"] / tab["acidentes_total"]  
    
    ### filtro opcional de volume mínimo 
    tab = tab[tab["acidentes_total"] >= min_acidentes] 
    
    return tab.sort_values("pct_acidente_fatal", ascending = False) 
```
